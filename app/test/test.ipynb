```python
from sqlalchemy import create_engine, inspect
import pandas as pd
```

```python
DB_USER = "root"
DB_PASSWORD = "Gayuth12345!"
DB_HOST = "localhost"
DB_PORT = "3306"
DB_NAME = "rental_db"

DATABASE_URL = (
    f"mysql+pymysql://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
)

engine = create_engine(DATABASE_URL)
engine.connect()
```

```python
inspector = inspect(engine)

tables = inspector.get_table_names()
tables
```

```python
schema_info = {}

for table in tables:
    columns = inspector.get_columns(table)
    schema_info[table] = [
        {
            "name": col["name"],
            "type": str(col["type"])
        }
        for col in columns
    ]

schema_info
```

```python
pd.read_sql("SELECT * FROM listing LIMIT 5;", engine)
```
